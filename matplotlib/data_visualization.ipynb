```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
car_df = pd.read_csv('../pandas/car-sales.csv')
car_df.head()
```

```python
car_df['Price'] = car_df['Price'].str.replace("[$,.]", "", regex=True).astype("Int64")
car_df["Price"] = car_df["Price"]/100
car_df["Price"]
```

```python
car_df["Sale Date"] =pd.date_range("10/01/2026", periods=len(car_df))
car_df
```

```python
car_df["Total Sale"] = car_df["Price"].cumsum()
car_df
```

```python
car_df.plot(x = "Sale Date", y="Total Sale");
```

```python
car_df.plot("Odometer (KM)", "Price", kind="scatter");
```

```python
car_df.plot.bar("Make", "Price");
```

```python
car_df['Price'].plot.hist(bins=20);
```
