```python
cars = pd.Series(['Audi','BMW','Ford'])
cars
```

```python
car_data = pd.DataFrame({
    "cars type" :cars,
    "color":colors
})
car_data
```

```python
food_type = pd.Series(['Apple','Orange','Banana'])
prices = pd.Series([2,3,4])
food_data = pd.DataFrame({
    "Food Type": food_type,
    "Price" : prices
})
food_data
```

# Converting string to int

```python
car_sales['Price'] = car_sales['Price'].str.replace('$','').str.replace(',' , '')
car_sales['Price'] = pd.to_numeric(car_sales['Price']).astype('int64')
car_sales['Price'].dtype
```
